```python
import json
from openai import OpenAI

ENDPOINT = "https://my-ai-projects.services.ai.azure.com/openai/v1"
DEPLOYMENT = "gpt-5.2"
API_KEY = ""

SYSTEM_PROMPT = """You are a customer support triage assistant. Classify the customer query into exactly one category and assign a priority level.

Categories: billing, technical issue, account issue, complaint
Priorities: low, medium, high

Priority guidelines:
- High: urgency words (urgent, ASAP, immediately), financial loss, service fully down, strong negative emotion
- Medium: moderate impact, partial service disruption, frustrated tone
- Low: routine requests, informational queries, no urgency

Respond with ONLY a JSON object:
{"category": "<category>", "priority": "<priority>", "reason": "<short explanation>"}

Examples:

User: "I was charged twice for my subscription and no one is responding. This is urgent!"
{"category": "billing", "priority": "high", "reason": "double charge and urgency mentioned"}

User: "My app keeps crashing when I try to upload files."
{"category": "technical issue", "priority": "medium", "reason": "application malfunction without urgency"}

User: "I want to change my email address linked to my account."
{"category": "account issue", "priority": "low", "reason": "routine account modification request"}"""
```

```python
def get_client():
    return OpenAI(base_url=ENDPOINT, api_key=API_KEY)
```

```python
def classify_query(query: str) -> dict:
    client = get_client()
    response = client.chat.completions.create(
        model=DEPLOYMENT,
        messages=[
            {"role": "system", "content": SYSTEM_PROMPT},
            {"role": "user", "content": query},
        ],
        response_format={"type": "json_object"},
        temperature=0.1,
    )
    return json.loads(response.choices[0].message.content)
```

```python
query = input("Enter customer query: ")
```

```python
result = classify_query(query.strip())
print(result)
```

```python
category = result.get("category", "unknown")
priority = result.get("priority", "unknown")
reason = result.get("reason", "")
```

```python
print(f"Category: {category}")
print(f"Priority: {priority}")
print(f"Reason: {reason}")
```
